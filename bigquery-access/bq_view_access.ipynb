# Querying an Authorized BigQuery View

Query the authorized view `wb-crisp-bean-1269.temporary_data.authorized_world_soccer_games` and display sample rows.

```python
import os

from google.cloud import bigquery

DATA_PROJECT = "wb-crisp-bean-1269"
BILLING_PROJECT = os.environ.get("GOOGLE_CLOUD_PROJECT", DATA_PROJECT)
DATASET = "temporary_data"
VIEW = "authorized_world_soccer_games"

client = bigquery.Client(project=BILLING_PROJECT)
```

```python
query = f"""
SELECT *
FROM `{DATA_PROJECT}.{DATASET}.{VIEW}`
LIMIT 10
"""

df = client.query(query).to_dataframe()
df
```
